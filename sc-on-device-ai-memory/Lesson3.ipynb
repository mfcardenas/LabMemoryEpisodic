# L3: Building the Memory

In this lab, you'll build a memory store from scratch using Qdrant Edge, and test how it stores, retrieves, filters, and forgets both text and images. By the end, you'll have a working multimodal memory store and a feel for how retrieval quality and speed change as you interact with it.

### Create an empty memory shard

A **shard** (also called a **collection**) is simply the directory where your memory store lives. Before you can store anything in it, you need to tell Qdrant Edge what kinds of vectors to expect.

Here, the shard is configured with two independent vector spaces:
- a **text** vector with 768 dimensions — the size produced by the text embedding model you'll use for notes
- an **image** vector with 512 dimensions — the size produced by CLIP, the model you'll use later for photos

Both use **cosine distance**, which compares vectors by their direction rather than their length (a common choice for embeddings) where direction is what carries meaning.

```python
from helper import *
from qdrant_edge import (Distance, EdgeConfig, EdgeShard,
                         EdgeVectorParams, Query, QueryRequest)

SHARD_DIR = fresh_start("./mem_shard")

config = EdgeConfig(
    vectors={
        "text": EdgeVectorParams(size=768, distance=Distance.Cosine),
        "image": EdgeVectorParams(size=512, distance=Distance.Cosine),
    }
)
shard = EdgeShard.create(SHARD_DIR, config)
print("Empty shard created at", SHARD_DIR)
```

### Querying an empty memory

Before loading anything in, it's worth checking that the shard is set up correctly. Try asking it any question and you'll should get back an empty result, since nothing has been stored yet.

The goal here is just to confirm the full path (embed → query → format results) works before you start adding real memories.

```python
question = "a good place to eat or drink nearby"

question_vector = embed_query(question)

hits = shard.query(
    QueryRequest(
        query=Query.Nearest(question_vector, using="text"),
        limit=3,
        with_payload=True,
    )
)
results_table(hits, "Closest memories", query=question)
```

## Load, embed, and store the notes

Now let's give the shard something to remember. The JSON file below contains a set of short text notes, each with some metadata (a **payload**) attached, things like a category, a price, a location, and a timestamp. This metadata won't affect how a note is retrieved by meaning, but it will let you filter results later.

```python
notes = load_memories(
    "./ro_shared_data/memories.json",
    source_type="text",
)
memories_table(notes[:6], f"First 6 of {len(notes)} text notes")
```

> Note: Each note gets converted into a vector using the same embedding model that will later encode your questions.

```python
vectors = embed_text([m["note"] for m in notes])

vector_preview(notes[0]["note"], vectors[0])
```

## Storing memories as points

In, Qdrant Edge, memories are stored as a **point**: a vector (or several, for multimodal memories) bundled with an ID and its metadata payload. Here, every note becomes a point holding its text vector alongside its `source_type`, `category`, `location`, `timestamp`, and `price`. After upserting the points, `optimize()` prepares the shard's index for fast searching.

```python
from qdrant_edge import Point, UpdateOperation

points = [
    Point(id=m["id"], vector={"text": vector}, payload=m)
    for m, vector in zip(notes, vectors)
]
shard.update(UpdateOperation.upsert_points(points))
shard.optimize()

print("Stored", shard.info().points_count, "memories")
point_card(points[0], shown=0)
```

Now, it is the time to ask the same question again, now that the shard actually has memories in it. You should see your top three matches ranked by similarity score (the higher the score), the closer that memory is to your question in meaning.

```python
question = "a good place to eat or drink nearby"

question_vector = embed_query(question)

hits = shard.query(
    QueryRequest(
        query=Query.Nearest(question_vector, using="text"),
        limit=3,
        with_payload=True,
    )
)
results_table(hits, "Closest memories", query=question)
```

## Recall with category and price filters

A similarity search alone can rank by meaning, but it can't guarantee a result also satisfies a hard requirement like being under a certain price. To combine both, you first need to build a **field index** on the metadata fields you want to filter on (without an index). That metadata is stored but can't be efficiently filtered.

```python
from qdrant_edge import PayloadSchemaType

shard.update(UpdateOperation.create_field_index(
    "category", PayloadSchemaType.Keyword))
shard.update(UpdateOperation.create_field_index(
    "price", PayloadSchemaType.Float))

shard.optimize()
print("Indexed category and price")
```

Now the same question is combined with a filter: only food-category memories under $15 should be returned. Notice that every result satisfies both the meaning-based similarity **and** the hard filter conditions. Filtering and similarity ranking work together rather than as substitutes for each other.

```python
from qdrant_edge import (FieldCondition, Filter, MatchValue,
                         RangeFloat)

cheap_food = Filter(
    must=[
        FieldCondition(key="category", match=MatchValue(value="food")),
        FieldCondition(key="price", range=RangeFloat(lt=15)),
    ]
)
hits_filtered = text_search(shard, question, cheap_food)

results_table(hits_filtered, "Food under $15", query=question)
```

## Adding images

Memory doesn't have to be text-only. In this step, we add a batch of photos to the same shard, using its image vector space. The embedding model behind this is **CLIP**, which maps both images and text into the same embedding space, that's what makes it possible to search photos using a plain text description instead of another image.

```python
store_photos(shard, "./ro_shared_data/bank")
```

Try describing an object in words here, for example, "a red bicycle" and see if the shard can find the matching photo. Because CLIP embeds text and images compatibly, the text query never needs to be turned into an image first.

```python
my_description = "a red bicycle"

description_vector = embed_query_clip(my_description)

photo_hits = shard.query(
    QueryRequest(
        query=Query.Nearest(description_vector, using="image"),
        limit=1,
        with_payload=True,
    )
)
show_photo_results(photo_hits, "./ro_shared_data/bank",
                  my_description)
```

## Forgetting

Not every memory should stick around forever. Forgetting here means explicitly deleting a specific point (its vector and its metadata) from the shard with `delete_points`. This is different from filtering it out of a single query: once deleted, that memory is gone from every future search.

```python
forgotten = hits[0]

shard.update(UpdateOperation.delete_points([forgotten.id]))
shard.optimize()

print(f"Forgot id {forgotten.id}: {forgotten.payload['note']}")
```

Compare the results before and after deleting the coffee-shop memory. Notice that the ranking shifts up beneath it, but the scores of the remaining memories don't change, deleting one memory doesn't affect how the others are scored, only which ones are available to return.

```python
hits_after = text_search(shard, question, limit=3)

display(results_table(hits, "Before", query=question))
results_table(hits_after, "After",
              caption="the other scores are unchanged")
```
