## Store the day

Now that everything is loaded, put it all into a single memory store the same way you did in the previous lesson (l3): create a shard, then embed and upsert each memory with its `source_type` (text, voice, or photo) attached as metadata.

```python
shard = new_shard("./day_shard", text=768, image=512)
```

>Note: Text notes and voice-note transcripts are stored the same way.

```python
texts = [m for m in memories if m["source_type"] in ("text", "voice")]
store_notes(shard, texts)
```

>Note: Photos are stored separately, into the image vector space you set up when creating the shard.

```python
photos = [m for m in memories if m["source_type"] == "photo"]
store_photo_memories(shard, photos, "./ro_shared_data/images")
```

## Recall your day

With everything stored, build a single `recall` function that searches across all three memory types at once: text notes, voice-note transcripts (which live in the same text vector space), and photos, and organizes the results by type.

```python
def recall(question):
    text_hits = text_search(shard, question, limit=10)
    photo_hits = photo_search(shard, question, limit=1)
    return {
        "Photos": photo_hits,
        "Voice Notes": [h for h in text_hits
                        if h.payload["source_type"] == "voice"][:3],
        "Text Notes": [h for h in text_hits
                       if h.payload["source_type"] == "text"][:3],
    }
```

```python
# Try asking a question that spans several of your stored memories

my_question = "what was the ramen place downtown"

hits = recall(my_question)
memory_inbox(hits, "./ro_shared_data/images",
             min_text_score=0.6, min_photo_score=0.23)
```

## Ask by voice

You don't have to type a question, you can ask out loud instead. The audio question gets transcribed with the same local Whisper model used for voice notes, and the resulting text is passed into the exact same `recall` function you just used.

```python
display(Audio("./ro_shared_data/audio/question.wav"))
print("Heard:", spoken)
```

This is the same audio-question flow: the transcribed question is passed straight into `recall`. Look closely at the notes that come back below the threshold line, you may see one that mentions "the park" without actually answering where the bike was parked. That's how the threshold works: a weak, coincidental wording match still gets retrieved, but stays grayed out instead of being surfaced as a real answer.

```python
hits = recall(spoken)
memory_inbox(hits, "./ro_shared_data/images",
             min_text_score=0.6, min_photo_score=0.23)
```

## Add a memory and recall it

Finally, test that a brand-new memory can be recalled immediately, with no extra setup. Add a new note the same way you stored notes earlier, embed it, wrap it in a `Point` with its metadata, and upsert it into the same shard alongside everything else.

```python
from qdrant_edge import Point, UpdateOperation

my_note = "Left the spare key with the neighbor in apartment 4B"

note_vector = embed_text([my_note])[0]

my_memory = Point(
    id=900,
    vector={"text": note_vector},
    payload={"source_type": "text", "category": "home",
             "location": "Home", "note": my_note},
)
shard.update(UpdateOperation.upsert_points([my_memory]))
```

Now ask a question about the memory you just added. If everything is wired up correctly, that new note should come back as the top match, with the highest similarity score, since it's a near word-for-word repeat of the question.

```python
my_question = "where did I leave the spare key"

hits = recall(my_question)
memory_inbox(hits, "./ro_shared_data/images",
             min_text_score=0.6, min_photo_score=0.23)
```
